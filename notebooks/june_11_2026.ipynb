```python
df["sex"] = df["sex"].astype("category")
df["embarked"] = df["embarked"].astype("category")
df["class"] = df["class"].astype("category")
df.dtypes
```

```python
df = df.drop(columns="deck")
df.columns
```

```python
numeric_columns = df.select_dtypes(include=["int64", "float64"]).columns
numeric_columns
```

```python
statistic_summary = pd.DataFrame()
statistic_summary["Mean"] = df[numeric_columns].mean()
statistic_summary["Median"] = df[numeric_columns].median()
statistic_summary["Mode"] = df[numeric_columns].mode().iloc[0]
statistic_summary["Variance"] = df[numeric_columns].var()
statistic_summary["Standard Deviation"] = df[numeric_columns].std()
statistic_summary["Q1"] = df[numeric_columns].quantile(0.25)
statistic_summary["Q3"] = df[numeric_columns].quantile(0.75)
statistic_summary["IQR"] = statistic_summary["Q3"] - statistic_summary["Q1"]
statistic_summary
```

*HYPOTHESIS TESTING*

```python
male_age = df[df["sex"] == "male"]["age"]
female_age = df[df["sex"]=="female"]["age"]
```

```python
print(male_age, female_age)
```

```python
from scipy.stats import ttest_ind

t_stats, p_value = ttest_ind(male_age, female_age, equal_var=False)

print("T-Statistic:", t_stats)
print("P-Value:", p_value)

alpha = .05

if p_value < alpha:
    print("Reject H0")
else:
    print("Fail to reject H0")
```

*Confidence Interval*

```python
sample_mean = df["fare"].mean()
print(f"Sample mean: {sample_mean:.2f}")
```

```python
sample_std = df["fare"].std()
sample_size = len(df)

standard_error = sample_std / np.sqrt(sample_size)

print(f"Standard Error: {standard_error}")
```

```python
lower_95 = sample_mean - 1.96 * standard_error
upper_95 = sample_mean + 1.96 * standard_error

print("95% Confidence Interval:\n")
print(lower_95, upper_95)
```

```python
lower_99 = sample_mean - 2.576 * standard_error
upper_99 = sample_mean + 2.576 * standard_error

print("95% Confidence Interval:\n")
print(lower_99, upper_99)
```
